```python
  # importando bibliotecas
import pandas as pd 
import numpy as np 
```

```python
    # lendo o arquivo CSV (limpo)
df = pd.read_csv('../dados/base_varejo_limpa.csv',sep=';')
```

```python
# analisando categorias vendidas por gênero 
agrupamento_genero = df.groupby(['PR_CAT', 'CL_GENERO']).agg(total_vendas=('CO_ID', 'count'))

print('\nCategoria x Gênero')
print(agrupamento_genero)
```

```python
# analisando produtos vendidos por gênero  
produto_genero = df.groupby(['PR_NOME', 'CL_GENERO']).agg(vendas=('CO_ID', 'count')).reset_index()

produto_genero = produto_genero.sort_values(by='vendas',ascending=False)

print(produto_genero.head(15))
```

```python
             #  estatísticas descritivas da coluna de número de filhos do cliente

    # média
media = df['CL_FHL'].mean()
print('Média:')
print(media)

    # mediana
mediana = df['CL_FHL'].median()
print('\nMediana:')
print(mediana)

    # desvio padrão
desvio = df['CL_FHL'].std()
print('\nDesvio padrão:')
print(desvio)

    # moda
moda = df['CL_FHL'].mode()
print('\nModa:')
print(moda)

    # valor máximo
maximo = df['CL_FHL'].max()
print('\nMáximo:')
print(maximo)

    # valor mínimo 
minimo = df['CL_FHL'].min()
print('\nMínimo:')
print(minimo)

    # contagem
contagem = df['CL_FHL'].count()
print('\nContagem:')
print(contagem)

    # quartis
Q1 = df['CL_FHL'].quantile(0.25)
Q2 = df['CL_FHL'].quantile(0.50)
Q3 = df['CL_FHL'].quantile(0.75)

print(f'Q1 (25%): {Q1}')
print(f'Q2 (50%): {Q2}')
print(f'Q3 (75%): {Q3}')
```
